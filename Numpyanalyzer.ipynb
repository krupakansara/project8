```python
#Create variable shared by all objects
class DataAnalytics:
    total_objects=0

    #constructor and encapsulation
    def __init__(self, array=None):
        if array is None:
            self.__array = np.array([])
        else:
            self.__array = np.array(array)
        DataAnalytics.total_objects += 1

    def __check_array(self):
        if self.__array.size==0:
            raise ValueError("No array available. Create an array first")

    def display_array(self):
        self.__check_array():
        self.__check_array()
        print("Current Array:")
        print(self.__array)
        print("Dimensions: ",self.__array.ndim)
        print("Shape: ", self.__array.shape)
        print("Total Elements: ",self.__array.size)

    #array creation
    def create_array(self):
        print("---Create array---")
        print("1. Crate 1D array")
        print("2. Create 2D array")
        print("3. Crate 3D array")

        choice = input("Enter your choice:")

        try:
            if choice == "1":
                values = input("Enter numbers seprated by commas: ")
                self.__array = np.array(
                    [float(x.strip()) for x in values.split(",")]
                )

            elif choice == "2":
                print("Enter each row on a new line")
                print("Separated numbers in a row using commas")
                print("type END when all rows are entered")

                rows=[]

                while True:
                    row=input("Enter row or END: ").strip()

                    if row.upper() == "END":
                        break

                    if row == "":
                        print("Row cannot be empty")
                        continue 

                    rows.append(
                        [float(x.strip()) for x in row.split(",")]
                    )

                    if len(rows) == 0:
                        print("No rows entered")
                        return

                    if not all(len(row) == len(rows[0]) for row in rows):
                        print("Error: All rows must have equal lenght")
                        return

                    self.__array = np.array(rows)

            elif choice == "3":
                depth = int(input("Enter depth: "))
                rows = int(input("Enter rows per matrix : "))
                columns = int(input("Enter columns per matrix: "))

                if depth <=0 or rows <= 0 or columns <=0:
                    print("Dimensions must be positive")
                    return 

                values = input(
                    "Enter all numbers seprated by commas: ")
                numbers = [float(x.strip()) for x in values.split(",")]

                if len(numbers) != depth*rows*columns:
                    print("incorrect number of elemets")
                    return

                self.__array = np.array(numbers).reshape(
                    depth,rows,columns
                )

            else:
                print("invalid choice")
                return

            print("Array created successfully")
            self.display_array()

        except ValueError as error:
            print("Invalid input: ",error)

    #indexing and slicing
    def indexing_slicing(self):
        try:
            self.__check_array()
            self.display_array()

            print("1. Access an element")
            print("2. Access a row")
            print("3. Access a column")
            print("4. Slice the array")

            choice = input("Enter your choice: ")

            if choice == "1":
                index = input(
                    "ENter idex(e.g. 2 or 1,2 or 0,1,2):"
                )
                indexes = tuple(int(x.strip()) for x in index.split(",")
                               )

                if len(indexes) == 1:
                    print("ELement:",self.__array[indexes[0]])
                else:
                    print("Element: ", self.__array[index])
                    
            elif choice == "2":
                if self.__array.ndim<2:
                    print("Rows require a 2D or 3D arrays")
                    return 

                index = int(input("Enter row index: "))
                print("Selected row: ", self.__array[index])

            elif choice == "3":
                if self.__array.ndim !=2:
                    print("Column selection requires a 2D array")
                    return 

                index = int(input("Enter column index: "))
                print("Selected column: ",self.__array[:,index])

            elif choice == "4":
                start = int(input("Enter starting index: "))
                stop = int(input("Enter ending index(exclusive):"))

                print("Sliced array:", self._-array[start:stop])

            else:
                print("Invalid choice")

        except (ValueError, IndexError, TypeError) as error:
            print("Unable to perform operation: ",error)

    def combine_arrays(self):
        try:
            self.__check_array()

            print("Current array shape: ", self.__array.shape)

            dimensions=self.__array.ndim
            shape=[]

            if dimensions == 1:
                length = int(input("Enter length of second array: "))
                shape = [length]

            elif dimensions == 2:
                rows = int(input("Enter rows of second arrat: "))
                columns = int(input("Enter columns of second array: "))
                shape = [rows, columns]

            elif dimensions == 3:
                depth = int(input("Enter depth of second array:"))
                rows=int(input("Enter rows per matrix:"))
                columns = int(input("Enter columns per matrix : "))
                shape = [depth, rows, columns]

            if any(size <=0 for size in shape):
                print("All dimensions must be positive")
                return

            total_elements = int(np.prod(shape))

            values = input("Enter" + str(total_elements) + "numbers separated by commas: ")

            numbers = [float(x.strip()) for x in values.split(",")]

            if len(numbers) != total_elements:
                print("Incoorect number of elements.")
                return

            second_array=np.array(numbers).reshape(shape)

            print("1. Concatenate along axis 0")
            if dimensions >=2:
                print("2. Concatenate along axis 1")
            if dimensions == 3:
                print("3. Concatenate along axis 2")

            axis = int(input("Enter axis numbers:"))

            if axis < 0 or axis >= dimensions:
                print("Invalid axis")
                return

            for i in range(dimensions):
                if i != axis and self.__array.shape[i] != second_array.shape[i]:
                    print("Arrays shapes are incompatible for this axis")
                    return

            combined = np.concatenate(
                (self.__array, second_array), axis=axis
            )

            print("Combined array:")
            print(combined)

        except (ValueError, TypeError, IndexError) as error:
            print("Unable to combine arrays: ",error)

    #Elements wise
    def mathematical_operations(self):
        try:
            self.__check_array()

            values = input("Enter array with compatible shape" "(comma seprated):"
                          )
            second_array = np.array([float(x.strip()) for x in values.split(",")]
                                   )
            if self.__array.shape != second_array.shape:
                print("Both arrays must have the same shape")
                return

            print("1. Addition")
            print("2. Subtraction")
            print("3. Multiplication")
            print("4. Division")

            choice = input("Enter your choice: ")

            if choice == "1":
                result = self.__array + second_array
            elif choice == "2":
                result = self.__array - second_array
            elif choice == "3":
                result = self.__array * second_array
            elif choice == "4":
                if np.any(second_array == 0):
                    print("Division by zero is not allowed")
                    return
                result = self.__array / second_array
            else:
                print("Invalid choice")
                return 

            print("Result:")
            print(result)

        except (ValueError, TypeError) as error:
            print("Unabel to perform operations: ",error)

    #dot product and matrix multiplication 
    def matrix_operations(self):
        try:
            self.__check_array()

            if self.__array.ndim != 2:
                print("The current array must be  20")
                return

            rows = int(input("Enter rows of the second matrix: "))
            columns = int(input("Enter columns of the second matrix: "))


            if rows <= 0 or columns <=0:
                print("Dimensions must be positive")
                return

            if self.__array.shape[1] != rows:
                print(
                    "Incompatible dimensions. The number of columns"
                    "in the first matrix must equal the number of rows"
                    "in the second matrix"
                )
                return

            values =  input("Enter the second matrix elements row by row,"
                            "separated by commas: "
                           )

            numbers = [float(x.strip()) for x in values.split(",")]

            if len(numbers) != rows * columns:
                print("Incorrect number of elements")
                return

            second_matrix = np.array(numbers).reshape(rows,columns)

            print("First matrix")
            print(self.__array)

            print("Second matrix")
            print(second_matrix)

            print("Dot product")
            print("Matrix multiplication")

            choice = input("Enter your choice: ")

            if choice == "1":
                result = np.dot(self._-array, second_matrix)
            elif choice == "2":
                result = np.matmul(self.___array, second_matrix)
            else:
                print("invalid choice")
                return 

            print("Result:")
            print(result)

        except (ValueError, TypeError) as error:
            print("Unable to perform matrix operations: ", error)

    #search, sort and filter
    def search_sort_filter(self):
        try:
            self.__check_array()

            print("1.Search for a value")
            print("2. Sort in ascending order")
            print("3. Sort in descending order")
            print("4. Filter using a condition")

            choice = input("Enter your choice:")

            if choice == "1":
                value=float(input("Enter value to search:"))
                positions = np.argwhere(self.__array == value)

                if positions.size==0:
                    print("Value not found")
                else:
                    print("Value found at index positions:")
                    print(positions)

            elif choice == "2":
                print("Sorted array: ")
                print(np.sort(self.__array, axis=None))

            elif choice == "3":
                print("Sorted array in descending order: ")
                print(np.sort(self.__array, axis=None)[::-1])

            elif choice == "4":
                print("Conditions: >,<.>=,<=,==,!=")
                operator = input("Enter conditions: ").strip()
                value = float("ENter comparision value: ")

                if operator == ">":
                    result = self.__array[self.__array > value]
                if operator == "<":
                    result = self.__array[self.__array < value]
                if operator == ">=":
                    result = self.__array[self.__array >= value]
                if operator == "<=":
                    result = self.__array[self.__array <= value]
                if operator == "==":
                    result = self.__array[self.__array == value]
                if operator == "!=":
                    result = self.__array[self.__array != value]
                else:
                    print("Invalid condition")
                    return

                print("Filtered values")
                print(result)

            else:
                print("Invaolid choice")
        except (ValueError, TypeError) as error:
            print("Unable to perform oepration:L ",error)

    #Aggregating functions
    def aggregation(self):
        try:
            self.__check_array()

            print("Sum: ",np.sum(self.__array))
            print("Mean: ", np.mean(self.__array))
            print("Median: ", np.mediam(self.__array))
            print("Standard Deviation: ",np.std(self.__array))
            print("Variamce: ", np.var(self.__array))

        except (ValueError, TypeError) as error:
            print("Unable to calculate statoistics :", error)

    #Statistical Functions
    def statistical_functions(self):
        try:
            self.__check_array()

            print("1. Minimum and maximum")
            print("2. Percentile")
            print("3. Correlation coefficient")

            choice = input("Enter your choice: ")

            if choice ==  "1":
                print("Minimum: ", np.min(self.__array))
                print("Maximum: ", np.max(self.__array))

            elif choice == "2":
                percentile = float(
                    input("Enter percentile (0 to 100): ")
                )

                if percentile < 0 or percentile > 100:
                    print("Percentile must be between 0 to 100")
                    return 

                print(
                    "Percentile value",
                    np.percentile(self.__array, percentile)
                )

            elif choice == "3":
                values = input("Enter another array of equal length"
                               "(comma-seprated): ")

                second_array = np.array(
                    [float(x.strip()) for x in values.split(",")]
                )

                first_array = self.__array.flatten()
                second_array = second_array.flatten()

                if first_array.size != second_array.size:
                    print("Both arrays must have equal lengths")
                    return

                if first_array.size < 2:
                    print("At least two values are requireed")
                    return

                if (np.std(first_array) == 0
                    or np.std(second_array) == 0):
                    print(
                        "Correlation is undefined when an array"
                        "has no variation"
                    )
                    return

                print(
                    "Correlation coefficinet: ",
                    np.corrcoef(first_array, second_array)[0,1]
                )
            else:
                print("Invalid choice")

        except (ValueError, TypeError) as error:
            print("Unable to calculate statistics: ", error)

    #privat computation and encapsulation
    def __calculate_summary(self):
        self.__check_array()

        return {
            "sum":np.sum(self.__array),
            "mean" : np.mean(self.__array),
            "minimum": np.min(self._-array),
            "maximum" : np.max(self._-array)
        }

    def show_summary(self):
        try:
            summary = self.__calculate_summary()

            print("Array Summary:")
            for key, value in summary.items():
                print(key.capitalize() + ":", value)

        except ValueError as error:
            print(error)

    #class method

    @classmethod
    def from_list(cls, values):
        return cls(values)

    @classmethod
    def show_object_count(cls):
        print("total DataAnalytics object created: ", cls.total_objects)

    @staticmethod
    def is_numeric(value):
        try:
            float(value)
            return True
        except (ValueError, TypeError):
            return False

    #Menu-driven
def main():
    analyzer = dataAnalytics()

    while True:
        print("\n" + "=" * 40)
        print("numpy Analyzer")
        print("=" * 40)
        print("1. Create an array")
        print("2. Display current array")
        print("3. Indexing and slicing")
        print("4. Combine arrays")
        print("5. Split array")
        print("6. Mathematical operations")
        print("7. Dot product and matrix multiplication")
        print("8. Search, sort and filter")
        print("9. Aggregation functions")
        print("10.Statistical functions")
        print("11. Display Array summary")
        print("12. Display object count")
        print("0. Exit")

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            analyzer.create_array()

        elif choice == "2":
            try:
                analyzer.display_array()
            except ValueError as error:
                print(error)

        elif choice == "3":
            analyzer.indexing_slicing()

        elif choice == "4":
            analyzer.combine_arrays()

        elif choice == "5":
            analyzer.split_array()

        elif choice == "6":
            analyzer.mathematical_operations()

        elif choice == "7":
            analyzer.matrix_operations()

        elif choice == "8":
            analyzer.search_sort_filter()

        elif choice == "9":
            analyzer.aggregation()

        elif choice == "10":
            analyzer.statistical_functions()

        elif choice == "11":
            analyzer.show_summary()

        elif choice == "12":
            DataAnalytics.show_object_count()

        elif choice == "0":
            print("Thank you for using Numpy Analyzer!")
            break

        else:
            print("Invalid choice. please try again")

if __name__ == "__main__":
    main()
    
 
        
            
        
                
                    
                    
                
                           
                

                          

                
        
        
```
